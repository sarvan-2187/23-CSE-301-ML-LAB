1. Create a quantum circuit with a single qubit and display the circuit diagram.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)

print(qc.draw())
```

2. Write a Qiskit program to apply the Pauli-X (NOT) gate on a qubit and observe the circuit.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)

qc.x(0)

print(qc.draw())
```

3. Create a quantum circuit using the Hadamard gate and study the concept of superposition.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)

qc.h(0)

print(qc.draw())
```
